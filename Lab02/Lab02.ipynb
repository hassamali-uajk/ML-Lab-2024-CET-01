# Task No. 06

```python
# [Task 6]
units = np.array(["Punjab", "Sindh", "KP", "Balochistan", "Islamabad", "AJK"])
pop_2017 = np.array([110.01, 47.85, 35.53, 12.34, 2.01, 4.05])
p23 = np.array([127.69, 55.70, 40.85, 14.89, 2.36, 4.05])

def mk_dict(turns):
    return {str(kar_idx): (x.item() if hasattr(x, "item") else x) for kar_idx, x in turns}

all_pop = np.sum(p23)
print(f"Q6a. total population 2023 = {all_pop:.2f} million")

share = 100 * p23 / all_pop
print(f"Q6b. share of total (%): {mk_dict(zip(units, np.round(share, 2)))}")

print(f"Q6c. most populous unit: {units[p23.argmax()]}")
print(f"Q6d. units above 20 million: {units[p23 > 20]}")

growth_rate = (np.power(p23 / pop_2017, 1 / 6) - 1) * 100
print(f"Q6e. annual growth 2017-2023 (% per year): {mk_dict(zip(units, np.round(growth_rate, 2)))}")

print(f"Q6f. fastest-growing unit: {units[np.argmax(growth_rate)]} ({growth_rate.max():.2f}% per year)")

national = ((p23.sum() / pop_2017.sum()) ** (1 / 6) - 1) * 100
years_to_double = 70 / growth_rate.max()
print(f"Q6 interpret. national growth = {national:.2f}% per year.")
print(f"Q6 interpret. {units[np.argmax(growth_rate)]} would double in about {years_to_double:.0f} years at its current rate.")
print(f"Q6 interpret. AJK grew {growth_rate[-1]:.2f}% per year vs national {national:.2f}% -> "
      f"AJK is {national - growth_rate[-1]:.2f} points below the national rate (in this sample AJK is basically flat).")
```

# Task No. 07

```python
# [Task 7]
towns = np.array(["Lahore", "Karachi", "Islamabad", "Quetta", "Peshawar"])
months = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                   "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])

temps = np.array([
    [19, 22, 28, 35, 40, 40, 36, 34, 34, 32, 26, 21],
    [26, 28, 32, 34, 35, 34, 32, 31, 32, 34, 31, 28],
    [17, 19, 24, 30, 36, 38, 35, 33, 33, 30, 24, 19],
    [11, 13, 18, 24, 29, 34, 35, 34, 31, 25, 19, 13],
    [18, 20, 25, 31, 37, 40, 38, 36, 35, 31, 25, 19],
])
rain_data = np.array([
    [12, 22, 34, 22, 21, 54, 214, 232, 82, 8, 3, 8],
    [8, 10, 9, 4, 1, 6, 40, 25, 13, 1, 2, 5],
    [56, 65, 76, 52, 32, 66, 260, 310, 120, 28, 12, 28],
    [38, 40, 33, 20, 8, 5, 18, 15, 5, 3, 7, 18],
    [33, 48, 63, 48, 23, 21, 86, 95, 43, 23, 20, 29],
])

def mk_dict(turns):
    return {str(kar_idx): (x.item() if hasattr(x, "item") else x) for kar_idx, x in turns}

print(f"Q7a. shape of temp (cities, months): {temps.shape}")

year_avg = np.mean(temps, axis=1)
print(f"Q7b. yearly mean temperature (deg C): {mk_dict(zip(towns, np.round(year_avg, 1)))}")

warmest = months[np.argmax(temps, axis=1)]
print(f"Q7c. hottest month per city: {mk_dict(zip(towns, warmest))}")

print(f"Q7d. 5-city monthly average (deg C): {mk_dict(zip(months, np.round(temps.mean(axis=0), 1)))}")

above35 = np.count_nonzero(temps > 35, axis=1)
print(f"Q7e. months above 35 deg C: {mk_dict(zip(towns, above35))}")

lahore_fahr = temps[0] * 9 / 5 + 32
print(f"Q7f. Lahore monthly temperature (deg F): {np.round(lahore_fahr, 1)}")

mons = rain_data[:, 6:9].sum(1)
print(f"Q7g. monsoon rain Jul-Sep (mm): {mk_dict(zip(towns, mons))}")
print(f"Q7g. wettest monsoon city: {towns[np.argmax(mons)]} ({mons.max()} mm)")

share = mons / rain_data.sum(axis=1) * 100
print(f"Q7h. monsoon share of annual rain (%): {mk_dict(zip(towns, np.round(share, 1)))}")

kar_idx = np.where(towns == "Karachi")[0][0]
print(f"Q7 interpret. Karachi has the highest yearly mean ({year_avg.max():.1f} deg C) "
      f"but {above35[kar_idx]} months above 35 deg C, while cooler-on-average Peshawar has {above35[4]} "
      f"and Lahore {above35[0]}. A high yearly mean does not mean the most extreme heat: "
      f"Karachi is hot all year, the inland cities spike in summer.")
```

# Task No. 08

```python
# [Task 8]
fy = np.array(["FY2023", "FY2024", "FY2025"])
fiscal_months = np.array(["Jul", "Aug", "Sep", "Oct", "Nov", "Dec",
                    "Jan", "Feb", "Mar", "Apr", "May", "Jun"])

cpi = np.array([
    [24.9, 27.3, 23.2, 26.6, 23.8, 24.5, 27.6, 31.5, 35.4, 36.4, 38.0, 29.4],
    [28.3, 27.4, 31.4, 26.9, 29.2, 29.7, 28.3, 23.1, 20.7, 17.3, 11.8, 12.6],
    [4.1, 9.6, 6.9, 7.2, 4.9, 4.1, 2.4, 1.5, 0.7, 0.3, 3.5, 3.2],
])

def mk_dict(turns):
    return {str(kar_idx): (x.item() if hasattr(x, "item") else x) for kar_idx, x in turns}

yr_avg = np.mean(cpi, axis=1)
print(f"Q8a. average inflation per year (%): {mk_dict(zip(fy, np.round(yr_avg, 2)))}")

fy24_row = cpi[1]
print(f"Q8b. FY2024 months above 25 %: {fiscal_months[fy24_row > 25]} -> {(fy24_row > 25).sum()} months")

change = np.diff(cpi, axis=1)
print(f"Q8c. month-on-month change (percentage points; negative = drop):\n{np.round(change, 1)}")

biggest_drop = change.argmin(axis=1)
print(f"Q8d. month of biggest drop: {mk_dict(zip(fy, fiscal_months[biggest_drop + 1]))} | size (pp): {np.round(change.min(axis=1), 1)}")

fy25_vals = cpi[2]
tags = np.where(fy25_vals < 5, "Low", np.where(fy25_vals < 10, "Moderate", "High"))
print(f"Q8e. FY2025 labels (Low < 5 %, Moderate 5-10 %, High >= 10 %):\n{np.column_stack((fiscal_months, fy25_vals, tags))}")

print(f"Q8 interpret. average inflation went {yr_avg[0]:.1f}% -> {yr_avg[1]:.1f}% -> {yr_avg[2]:.1f}%. "
      f"Prices did NOT fall; they kept rising, but more and more slowly (disinflation), "
      f"so the price LEVEL is still much higher than at the start of FY2023.")
```

# Task No. 09

```python
# [Task 9]
student_marks = np.array([34, 38, 35, 52, 67, 73, 88, 95, 98, 39,
                  41, 45, 36, 58, 62, 71, 80, 33, 99, 44])
rollno = np.arange(1, 21)
PASS_MARK = 40

graced = np.clip(5 + student_marks, 0, 100)
print(f"Q9a. original marks: {student_marks}")
print(f"Q9a. marks after +5 grace (cap 100): {graced}")

p_before = (student_marks >= PASS_MARK).mean() * 100
after_pct = 100 * np.mean(graced >= PASS_MARK)
print(f"Q9b. pass % before grace = {p_before:.1f} %, after grace = {after_pct:.1f} %")

only_grace = (student_marks < PASS_MARK) & (graced >= PASS_MARK)
print(f"Q9c. passed only because of grace (roll numbers): {rollno[only_grace]}")
print(f"Q9c. their original marks: {student_marks[only_grace]}")

zscores = (graced - np.mean(graced)) / np.std(graced)
print(f"Q9d. mean = {graced.mean():.2f}, std = {graced.std():.2f}")
print(f"Q9d. z-scores (std devs from mean): {np.round(zscores, 2)}")
```

```python
# [Task 10]
pixels = np.zeros((8, 8), dtype=np.uint8)
pixels[1, 1:-1] = 255
pixels[[2, 3, 4, 5, 6], [6, 5, 5, 4, 4]] = 255

def view01(a):
    return (a > 0).astype(int)

print(f"Q10a. the '7' (1 = white):\n{view01(pixels)}")

flip_img = np.fliplr(pixels)
print(f"Q10b. flipped left-right:\n{view01(flip_img)}")

inverted = ~pixels
print(f"Q10c. inverted (white <-> black):\n{view01(inverted)}")

crop = pixels[0:7, 1:8]
print(f"Q10d. cropped 7x7 region, shape: {crop.shape}\n{view01(crop)}")

img_bright = np.clip(pixels.astype(int) + 60, 0, 255).astype(np.uint8)
print(f"Q10e. brightened (+60, capped at 255); unique values: {np.unique(img_bright)}")

print(f"Q10f. shape = {pixels.shape} | dtype = {pixels.dtype} | white pixels = {(pixels == 255).sum()}")
```
