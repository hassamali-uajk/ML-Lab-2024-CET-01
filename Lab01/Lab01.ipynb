Task no 10:-

```python
def check_speeding(current_speed, birthday):

    if birthday:
        current_speed -= 5

    if current_speed <= 60:
        return "No ticket"
    elif current_speed <= 80:
        return "small ticket"
    else:
        return "big ticket"

print("Q10", f"{check_speeding(60, False)}, {check_speeding(65, True)}, {check_speeding(85, False)}")
```
